```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, train_test_split
```

```python
df_train_traffic = pd.read_csv('../../data/df_train_traffic_encoded_scaled.csv')
df_train_average_bill = pd.read_csv('../../data/df_train_average_bill_encoded_scaled.csv')
df_train_traffic.head()
```

```python
def fitted_rf_model_best(df_train: pd.DataFrame, params):

    if 'Трафик' in df_train.columns:
        y = df_train['Трафик']
        X = df_train.drop(columns=['Трафик'])
    if 'Средний чек' in df_train.columns:
        y = df_train['Средний чек']
        X = df_train.drop(columns=['Средний чек'])

    X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.3, random_state=598)

    grid_search = GridSearchCV(
        RandomForestRegressor(random_state=598),
        params,
        cv=5,
        scoring = 'neg_root_mean_squared_error', 
        n_jobs=-1
    )

    return grid_search.fit(X_valid, y_valid)
```

```python
param_grid_1 = {
        'n_estimators': [200],
        'max_depth': [10, 15, 20],
        'max_features': [0.3, 0.5, 0.7],
        'max_samples': [0.7, 0.8, 0.9, 1.0]
    }

best_traffic_model = fitted_rf_model_best(df_train_traffic, param_grid_1)

print(f'Лучшая модель RF для предсказания трафика с параметрами: {best_traffic_model.best_estimator_}')
print(f'neg_root_mean_squared_error для предсказания трафика: {best_traffic_model.best_score_}')

best_average_bill_model = fitted_rf_model_best(df_train_average_bill, param_grid_1)

print(f'Лучшая модель RF для предсказания среднего чека с параметрами: {best_average_bill_model.best_estimator_}')
print(f'neg_root_mean_squared_error для предсказания среднего чека: {best_average_bill_model.best_score_}')
```
