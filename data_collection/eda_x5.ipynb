# **1. Описание исходных данных**

**new_id** - (числовой идентификатор) Уникальный ID магазина/точки                                                                         

**Месяц** - (числовой, 1–12) Номер месяца наблюдения   

**Трафик** - (числовой) Основной показатель посещаемости/потока клиентов магазина за месяц. Количество чеков за период

**Средний чек**- (числовой) Средняя сумма покупки в магазине за месяц

**Дата открытия, категориальный** - (категориальный) Возраст магазина в виде категории. В данных есть значения: Новый, Средний по возрасту, Открыт давно. Показывает, насколько давно работает магазин                                                        

**Торговая площадь, категориальный** - (категориальный) Размер магазина по торговой площади. В данных есть категории: Маленький, Средний, Большой, Очень большой

**Населенный пункт** - (текстовый) Город, поселок, село или другой населенный пункт, где расположен магазин

**Регион** - (текстовый) Регион РФ, где находится магазин.

**Численность населения** - (числовой) Население населенного пункта, где расположен магазин

**Количество домохозяйств** - (числовой) Количество домохозяйств в зоне/населенном пункте. Показывает потенциальную клиентскую базу                                                                                                                       
**Трафик пеший, в час** - (числовой) Средний пеший поток рядом с магазином за час. Чем выше показатель, тем больше потенциальных покупателей проходит мимо

                                                                        
**Трафик авто, в час** - (числовой) Средний автомобильный поток рядом с магазином за час. Важен для точек около дорог, парковок, транспортных узлов

**Маркетплейсы, доставки, постаматы (100 м)** - (числовой) Количество объектов доставки, маркетплейсов и постаматов в радиусе 100 метров

**Медицинские уч. и аптеки (300 м)** - (числовой) Количество медицинских учреждений и аптек в радиусе 300 метров. Это фактор инфраструктуры рядом с магазином

**Школы (300 м)** - (числовой) Количество школ в радиусе 300 метров. Может влиять на дневной поток и спрос в районе

**Остановки (300 м)** - (числовой) Количество остановок общественного транспорта в радиусе 300 метров. Важный показатель транспортной доступности

**Продуктовые магазины (500 м)** - (числовой) Количество продуктовых магазинов-конкурентов в радиусе 500 метров. Показывает уровень конкуренции в локации

**Пятерочки (500 м)** - (числовой) Количество других магазинов «Пятерочка» в радиусе 500 метров. Это внутренняя конкуренция/каннибализация внутри сети

Уже здесь видны признаки, которые требуют отдельной проверки на выбросы(трафик, средний чек, численность населения, пеший и автомобильный трафик) Также есть нулевые значения в некоторых признаках, например в численности населения, среднем чеке или инфраструктурных показателях, нужно проверить отдельно.

```python
df_before_outliers = df.copy()

df = df[
    (df["Средний чек"] >= 100) &
    (df["Численность населения"] >= 20) &
    (df["Количество домохозяйств"] >= 5) &
    (df["Трафик пеший, в час"] > 0) &
    (df["Трафик авто, в час"] > 0) &
    (df["Трафик"] > 0)
].copy()

print("Строк до удаления:", df_before_outliers.shape[0])
print("Строк после удаления:", df.shape[0])
print("Удалено строк:", df_before_outliers.shape[0] - df.shape[0])

print("Магазинов до удаления:", df_before_outliers["new_id"].nunique())
print("Магазинов после удаления:", df["new_id"].nunique())
print("Удалено магазинов:", df_before_outliers["new_id"].nunique() - df["new_id"].nunique())
```

**2.2. Проверка корректности месяцев по магазинам**

```python
df_original = df.copy()

correct_store_ids = []
incorrect_store_ids = []

for store_id, store_data in df.groupby("new_id"):
    months = sorted(store_data["Месяц"].tolist())
    expected_months = list(range(min(months), max(months) + 1))

    if months == expected_months and max(months) <= 12:
        correct_store_ids.append(store_id)
    else:
        incorrect_store_ids.append(store_id)

print("Корректных магазинов:", len(correct_store_ids))
print("Некорректных магазинов:", len(incorrect_store_ids))
```

Также была проведена проверка корректности месячных наблюдений по каждому магазину. В данных должны остаться только те магазины, у которых месяцы идут строго последовательно до n, где n не больше 12. Магазины с неполным, но корректным набором месяцев не удаляются

**2.3. Удаление магазинов с некорректными месяцами**

```python
df = df[df["new_id"].isin(correct_store_ids)].copy()

print("Было строк:", df_original.shape[0])
print("Стало строк:", df.shape[0])
print("Удалено строк:", df_original.shape[0] - df.shape[0])

print("Было магазинов:", df_original["new_id"].nunique())
print("Стало магазинов:", df["new_id"].nunique())
print("Удалено магазинов:", len(incorrect_store_ids))
```

Посмотрим на удаленные магазины

```python
incorrect_months = []

for store_id in incorrect_store_ids:
    months = sorted(df_original[df_original["new_id"] == store_id]["Месяц"].tolist())
    incorrect_months.append([store_id, months])

incorrect_months = pd.DataFrame(incorrect_months, columns=["new_id", "months"])
incorrect_months.head(20)
```

После проверки из данных были удалены магазины с некорректным набором месяцев: с пропусками внутри периода, дублями месяцев или месяцами, которые начинаются не с 1. Это сделано до основного EDA, чтобы дальнейший анализ строился только на корректных временных рядах

```python
months_by_store = df.groupby("new_id")["Месяц"].nunique().reset_index()
months_by_store = months_by_store.rename(columns={"Месяц": "months_count"})


full_year_store_ids = months_by_store[months_by_store["months_count"] == 12]["new_id"]
df_full_year = df[df["new_id"].isin(full_year_store_ids)].copy()

stores_full = df_full_year.sort_values("new_id")

print("Всего корректных магазинов:", df["new_id"].nunique())
print("Магазинов с полным годовым циклом:", df_full_year["new_id"].nunique())
print("Строк в df_full_year:", df_full_year.shape[0])
```

После очистки была рассчитана месячная выручка как произведение трафика на средний чек. Также отдельно выделены магазины с полным годовым циклом, то есть с наблюдениями за все 12 месяцев.

Дальше показатели, которые не связаны напрямую с сезонностью, рассматриваются на уровне уникальных магазинов с полным годовым циклом. Это позволяет не смешивать реальную структуру магазинов с техническим количеством строк.

**3.8. Проверка различий значений внутри одного населенного пункта**

```python
stores_for_city = (
    df
    .sort_values("Месяц")
    .drop_duplicates("new_id")
)
```

```python
city_value_check = (
    stores_for_city
    .groupby("Населенный пункт")
    .agg(
        stores_count=("new_id", "count"),
        households_unique=("Количество домохозяйств", "nunique"),
        foot_traffic_unique=("Трафик пеший, в час", "nunique"),
        auto_traffic_unique=("Трафик авто, в час", "nunique")
    )
    .reset_index()
)

city_value_check.head()
```

```python
cities_with_different_values = city_value_check[
    (city_value_check["households_unique"] > 1) |
    (city_value_check["foot_traffic_unique"] > 1) |
    (city_value_check["auto_traffic_unique"] > 1)
]

cities_with_different_values.sort_values("stores_count", ascending=False).head(20)
```

Проверка показывает, что внутри одного и того же населенного пункта могут встречаться разные значения по количеству домохозяйств, пешему и автомобильному трафику. Значит, эти признаки не являются полностью фиксированными на уровне города и могут отличаться для разных магазинов внутри одного населенного пункта.

```python
city_fill_values = stores_for_city.groupby("Населенный пункт").agg({
    "Количество домохозяйств": "median",
    "Трафик пеший, в час": "median",
    "Трафик авто, в час": "median"
}).reset_index()

city_fill_values.head()
```

Для дальнейшего обогащения спарсенного датасета по населенному пункту можно брать наиболее часто встречающееся значение признака внутри конкретного города. Такой подход выбирает типичное значение и не зависит от случайного магазина.

# **5. Первичная статистика по месяцам для магазинов с полным годовым циклом**

```python
monthly_stats_full = df_full_year.groupby("Месяц").agg(
    stores_count=("new_id", "nunique"),
    traffic_sum=("Трафик", "sum"),
    revenue_sum=("Выручка", "sum"),
    avg_check_mean=("Средний чек", "mean"),
    avg_check_median=("Средний чек", "median")
).reset_index()

monthly_stats_full["avg_check_weighted"] = (
    monthly_stats_full["revenue_sum"] / monthly_stats_full["traffic_sum"]
)

monthly_stats_full.round(2)
```

Для месячной динамики использованы только магазины с полным годовым циклом. В этом случае количество магазинов одинаковое в каждом месяце, поэтому суммарный трафик и суммарная расчетная выручка отражают сезонную динамику, а не изменение числа строк в данных.

Достаточно хорошо отслеживается тенденция, что в холодное время года трафик заметно падает. Средний чек резко увеличивается в декабре, но самая сильная сезонность прослеживается в выручке, даже несмотря на то что трафик в холодное время года уменьшается максимум выручки приходится именно на декабрь. Таким образом можно сделать вывод, что сезонность действительно прослеживается

# **5.1. Динамика суммарного трафика по месяцам**

```python
fig = px.line(
    monthly_stats_full,
    x="Месяц",
    y="traffic_sum",
    markers=True,
    title="Суммарный трафик по месяцам для магазинов с полным годовым циклом",
    template="plotly_white"
)

fig.show()
```

Суммарный трафик по месяцам теперь рассчитан только по магазинам с полным годовым циклом. Поэтому график можно интерпретировать как сезонную динамику трафика, а не как эффект разного количества магазинов в разные месяцы.

График подтвержает выводы про сезонность трафика: самый высокий трафитк приходится на лето (стоит предположить, что это связано с жарой и потребности в прохладительных напитках и снеков в виде мороженого во время летних прогулок). Далее виден резкий спад к январю - предполоительно люди доедают остатки новогоднего стола и менее нужнаются в походах в магазин, также из-за еще более неблагоприятных погодных условий скорее всего проще заказать доставку продуктов на дом

Будем объединять по следубщему принципу:

**Гипотеза 5: регионы отличаются по годовой выручке**
