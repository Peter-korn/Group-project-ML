```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

KNN для среднего чека

```python
df_average_bill = pd.read_csv("../../data/df_train_average_bill_encoded_scaled.csv")
df_average_bill.head()
```

```python
X = df_average_bill.drop("Средний чек", axis=1)
y = df_average_bill["Средний чек"]

print("Размер X:", X.shape)
print("Размер y:", y.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, random_state=56)

print("Train:", X_train.shape)
print("Test:", X_val.shape)
```

```python
knn = KNeighborsRegressor(n_jobs=-1)

param_grid = {
    "n_neighbors": list(range(1, 20))
}

gridsearch = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    scoring="neg_mean_absolute_error",
    cv=5,
    n_jobs=-1
)

gridsearch.fit(X_val, y_val)
```

```python
print("Лучшие параметры:", gridsearch.best_params_)
print("Лучшее MAE на кросс-валидации:", -gridsearch.best_score_)
```

```python
knn_average_bill_best = gridsearch.best_estimator_
```
