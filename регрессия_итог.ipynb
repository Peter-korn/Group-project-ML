```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
ALPHAS = [0.001, 0.01, 0.1, 1, 10, 100]
```

```python
df_traffic = pd.read_csv("new_traffic.csv")
df_average_bill = pd.read_csv("new_average.csv")

df_traffic.shape, df_average_bill.shape
```

```python
def train_ridge(df, target):

    X = df.drop(columns=[target])
    y = df[target]

    X_train, X_val, y_train, y_val = train_test_split(
        X, y,
        test_size=0.3,
        random_state=598
    )

    grid = GridSearchCV(
        Ridge(),
        {"alpha": ALPHAS},
        scoring="neg_root_mean_squared_error",
        cv=5,
        n_jobs=-1
    )

    grid.fit(X_val, y_val)

    return {
        "Целевая переменная": target,
        "Лучший alpha": grid.best_params_["alpha"],
        "RMSE": -grid.best_score_
    }
```

```python
traffic_result = train_ridge(df_traffic, "Трафик")
average_bill_result = train_ridge(df_average_bill, "Средний чек")
```

```python
results = pd.DataFrame([traffic_result, average_bill_result])
results
```
